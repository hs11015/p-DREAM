```python
import os
import pandas as pd

dir_path = "/data/DAICWOZ/EATD-Corpus"

# Training list
t_list = ['t_1', 't_103', 't_104', 't_105', 't_10', 't_107', 't_101', 't_102', 't_111', 't_106', 't_13', 't_11', 't_15', 't_109', 't_12', 't_2', 't_14', 't_16', 't_26', 't_18', 't_27', 't_20', 't_29', 't_21', 't_30', 't_22', 't_32', 't_28', 't_34', 't_3', 't_36', 't_31', 't_39', 't_33', 't_35', 't_42', 't_43', 't_44', 't_38', 't_46', 't_40', 't_48', 't_5', 't_51', 't_41', 't_54', 't_45', 't_56', 't_47', 't_53', 't_6', 't_55', 't_62', 't_58', 't_64', 't_63', 't_67', 't_59', 't_69', 't_60', 't_70', 't_61', 't_76', 't_78', 't_66', 't_8', 't_68', 't_75', 't_85', 't_86', 't_79', 't_88', 't_82', 't_9', 't_81', 't_91', 't_87', 't_89', 't_97', 't_99', 't_90', 't_93', 't_96']
# Validation list
v_list = ['v_10', 'v_103', 'v_106', 'v_12', 'v_104', 'v_15', 'v_105', 'v_17', 'v_11', 'v_19', 'v_13', 'v_20', 'v_21', 'v_22', 'v_16', 'v_24', 'v_18', 'v_2', 'v_28', 'v_29', 'v_23', 'v_31', 'v_32', 'v_36', 'v_37', 'v_25', 'v_39', 'v_27', 'v_43', 'v_30', 'v_4', 'v_47', 'v_41', 'v_49', 'v_42', 'v_51', 'v_52', 'v_45', 'v_55', 'v_48', 'v_56', 'v_5', 'v_62', 'v_63', 'v_53', 'v_65', 'v_66', 'v_67', 'v_61', 'v_7', 'v_69', 'v_57', 'v_73', 'v_58', 'v_74', 'v_64', 'v_75', 'v_70', 'v_77', 'v_72', 'v_79', 'v_71', 'v_81', 'v_82', 'v_76', 'v_8', 'v_88', 'v_83', 'v_89', 'v_84', 'v_86', 'v_96', 'v_94', 'v_87', 'v_9', 'v_90', 'v_98', 'v_99', 'v_97']


def get_label_info(subject_list, dataset_type):
    records = []
    for subject_id in subject_list:
        subject_dir = os.path.join(dir_path, subject_id)
        label_path = os.path.join(subject_dir, "new_label.txt")
        
        if os.path.exists(label_path):
            with open(label_path, 'r') as f:
                try:
                    score = float(f.read().strip())
                    status = "Depression" if score >= 53 else "Normal"
                    records.append({
                        "id": subject_id,
                        "score": score,
                        "label": status,
                        "set": dataset_type
                    })
                except ValueError:
                    print(f"Invalid value in {label_path}")
        else:
            print(f"Label file not found: {label_path}")
    
    return records

# Collect data from training and validation
train_data = get_label_info(t_list, "train")
valid_data = get_label_info(v_list, "valid")

# Combine into a single DataFrame
df = pd.DataFrame(train_data + valid_data)

# (Optional) sort by id
df = df.sort_values("id").reset_index(drop=True)

# Save as CSV if needed
df.to_csv(f"{dir_path}/label_table.csv", index=False)

print(df.head())
```

### Data augementation. 6개 조합

```python
import os
import itertools
import soundfile as sf
import numpy as np

# 기본 경로
base_dir = '/data/DAICWOZ/EATD-Corpus'

# 대상 감정
emotions = ['positive', 'negative', 'neutral']

# 모든 subject 디렉토리 순회
for subject in os.listdir(base_dir):
    subject_path = os.path.join(base_dir, subject)
    if not os.path.isdir(subject_path):
        continue  # 디렉토리가 아니면 건너뜀

    # 각 감정별 파일 경로 확인
    emotion_files = {}
    for emo in emotions:
        file_path = os.path.join(subject_path, f"{emo}.wav")
        if os.path.exists(file_path):
            emotion_files[emo] = file_path
        else:
            print(f"{subject}에 {emo}.wav 없음 [SKIP]")
            break
    else:
        # 모든 감정 파일이 존재하면 조합 시작
        for order in itertools.permutations(emotions):
            new_name = '_'.join([emo[:3] for emo in order]) + '.wav'
            output_path = os.path.join(subject_path, new_name)

            if os.path.exists(output_path):
                print(f"이미 존재: {output_path} [SKIP]")
                continue

            # 오디오 읽고 concat
            try:
                audio_parts = []
                sr = None
                for emo in order:
                    audio, sr = sf.read(emotion_files[emo])
                    audio_parts.append(audio)
                combined = np.concatenate(audio_parts, axis=0)
                sf.write(output_path, combined, sr)
                print(f"생성 완료: {output_path}")
            except Exception as e:
                print(f"생성 실패: {output_path} - {e}")
```
