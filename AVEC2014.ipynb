```python
import os
from glob import glob
from tqdm import tqdm
from moviepy import VideoFileClip

# Freeform directory
freeform_dir = "/data/DAICWOZ/AVEC2014/test/Northwind"  # train, dev, test 중 하나씩 반복
output_ext = ".wav"

# mp4 file list 
mp4_files = sorted(glob(os.path.join(freeform_dir, "*.mp4")))

print(f"Found {len(mp4_files)} mp4 files.")

# mp4 → wav transformation loop
for mp4_path in tqdm(mp4_files, desc="Converting MP4 to WAV"):
    try:
        wav_path = mp4_path.replace(".mp4", output_ext)
        if os.path.exists(wav_path):
            continue  # skip when already transformed
        clip = VideoFileClip(mp4_path)
        clip.audio.write_audiofile(wav_path, codec='pcm_s16le', fps=16000)
        clip.close()
    except Exception as e:
        print(f"❌ Error converting {mp4_path}: {e}")
```
